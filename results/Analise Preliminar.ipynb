```python
import pandas as pd
```

```python
df = pd.read_csv("iotlab1/1071-Number_Spiral-c++-mult-0W-iotlab1-20-02-2026-16-23.csv")
```

```python
def calc_tempo_relogio (df):
    tempo_relogio = sum(df["duration_time"].tolist())
    print(f"{tempo_relogio / 10**9:.2f} segundos")
    return tempo_relogio
```

```python
def calc_tempo_cpu (df):
    tempo_cpu = df["user_time"].tolist() + df["system_time"].tolist()
    tempo_cpu = sum(tempo_cpu)
    print(f"{tempo_cpu / 10**9:.2f} segundos")
    return tempo_cpu
```

```python
def calc_soma_cols (df, *cols):
    values = []
    for x in cols:
        values += df[x].tolist()
    return sum(values)
```

```python
def calc_taxa_cpu_relogio (df):
    return calc_tempo_cpu(df) / calc_tempo_relogio(df)
```
